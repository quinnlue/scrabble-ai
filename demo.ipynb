```python
from scrabble import Scrabble
from scrabble_ai import ScrabbleAI

game = Scrabble()
ai = ScrabbleAI(game)  # builds word graph (<5 seconds)
```

```python
from gui import SearchSettings

"""
max_depth: turns to look ahead
    - 1 means a greedy policy
    - I highly suggest not going beyond 3. The algorithm is designed for 3 turns of lookahead.
width: random opponent racks sampled per candidate
    - basically a lever to control the variance of the search
    - greater width -> lower variance
num_words_explored: candidate words tried at each ply
    - this is the number of words that are explored at each ply
    - basically we prune leaves that are not in the top N words by score
    - greater num_words_explored -> lower variance
random_opponent_rack: False = peek at the opponent's real rack
    - Use True if you want to simulate a realistic game under imperfect information
    - Use False if you want players to know their opponent's rack
"""


settings = SearchSettings(
    max_depth=2,
    width=5,
    num_words_explored=[8, 4],
    random_opponent_rack=True,
)
```

```python
from gui import ScrabbleGUI

ScrabbleGUI(game, ai, settings)
```
